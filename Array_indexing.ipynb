#### Numpy Indexing and slicing of Matrix

```python
arr = np.arange(1,31).reshape(6,5)
arr
```

```python
arr[0]
```

```python
arr[0,0]
```

```python
slice = arr[0:2,1:3]
slice
```

```python
arr[3:6,3:5]
```

```python
colum = arr[:,3]
colum
```

#### Boolean Indexing

```python
arr = np.arange(11,21)
arr
```

```python
bool_index = arr %2 == 0
bool_index
```

```python
arr = arr[bool_index]
arr
```
