#### **Array generation function**

**np.arange(strt,end)** => this function creates array of numbers given between these ranges 

```python
arr = np.arange(1,11)
arr
```

**np.zeros((row , colm))** => creates two dimension matrix of zeros(0) with the given rows and coloumns

```python
arr = np.zeros((4,3))
arr
```

**np.ones((row , colm))** => creates two dimension matrix of one(1) with the given rows and coloumns

```python
arr = np.ones((4,4))
arr
```

**np.linspace()** -> Linear space between range , 
linspace(1,5,3) => from 1 to 5 with  linear space gives 3 numbers 

```python
arr = np.linspace(1,5,3)
arr
```

```python
arr = np.linspace(1,10,50)
arr
```

#### **Random Generation Function**

```python
arr = np.random.rand(5)
arr
```

**np.random.rand()** => random.rand() function generates arrays of random number between 1 to 0 -> **Normalization**
but 
random.randn() function creates arrays of random number between -3 to 3 called **Statardisation**

```python
arr = np.random.randn(10)
arr
```

```python
arr = np.random.randint(1,10,10)
arr
```
